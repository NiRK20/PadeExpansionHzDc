Chaves dos dicionários de resultado:

`params`: contém os parâmetros do modelo em letras minúsculas. Cada parâmetro possui as chaves
    `mean`, `err_plus_1s`, `err_minus_1s`, `err_plus_2s`, `err_minus_2s`.

`logZ`: possui chaves `mean` e `err`

`BIC`: possui chaves `value`, `chi2min`, `n_data` e `n_pars`.

`nlives`: número de nlives usado no pypolychord.

`tempo`: tempo para rodar o modelo.

```python
from pandas import json_normalize
import json
import numpy as np
from IPython.display import display, IFrame

path_res = {
    'PS': {              
        'P21': {'path': './resultados/P21/PS/resultado_P21_PS.json', 'plot': './resultados/P21/PS/contornos_PS_P21.pdf', 'modelo': 'P21', 'dados': 'Pantheon+&SH0ES'},
        'P22': {'path': './resultados/P22/PS/resultado_P22_PS.json', 'plot': './resultados/P22/PS/contornos_PS_P22.pdf', 'modelo': 'P22', 'dados': 'Pantheon+&SH0ES'},
        'P31': {'path': './resultados/P31/PS/resultado_P31_PS.json', 'plot': './resultados/P31/PS/contornos_PS_P31.pdf', 'modelo': 'P31', 'dados': 'Pantheon+&SH0ES'},
        'P32': {'path': './resultados/P32/PS/resultado_P32_PS.json', 'plot': './resultados/P32/PS/contornos_PS_P32.pdf', 'modelo': 'P32', 'dados': 'Pantheon+&SH0ES'}
    }
}
```

```python
def extract_results(res_dict):
    for i in res_dict['params']:
        print(f'{i} = {np.round(res_dict["params"][i]["mean"], 3)} +{np.round(res_dict["params"][i]["err_plus_1s"], 3)} -{np.round(res_dict["params"][i]["err_minus_1s"], 3)} +{np.round(res_dict["params"][i]["err_plus_2s"], 3)} -{np.round(res_dict["params"][i]["err_minus_2s"], 3)}')

    print(f'\nlogZ = {np.round(res_dict["logZ"]["mean"], 2)} +/-{np.round(res_dict["logZ"]["err"], 2)}')

    print(f'\nBIC = {np.round(res_dict["BIC"]["value"], 2)}\n')

def comparar_modelos(mod_ref, mod_alv):
    print(f'Delta BIC = {np.round(mod_alv["BIC"]["value"] - mod_ref["BIC"]["value"], 3)}')
    print(f'Delta lnZ = {np.round(mod_alv["logZ"]["mean"] - mod_ref["logZ"]["mean"], 3)}')
```

```python
for i in path_res:
    for j in path_res[i]:
        with open(path_res[i][j]['path']) as f:
            print(f'Resultados do modelo {path_res[i][j]["modelo"]} para os dados do {path_res[i][j]["dados"]}:\n')
            extract_results(json.load(f))

        display(IFrame(path_res[i][j]['plot'], 815, 845))
        
        print('\n' + 120*'-' + '\n')
```

```python
with open(path_res['PS']['P21']['path']) as f:
    mod_ref_PS = json.load(f)

print('\n\nAnálise com o conjunto CC+Pantheon+&SH0ES+BAO')
print('\n' + 60*'-' + '\n')
for j in path_res['PS']:
    with open(path_res['PS'][j]['path']) as f:
        print(f'Comparando o modelo {path_res['PS'][j]["modelo"]} com P21:\n')
        comparar_modelos(mod_ref_PS, json.load(f))
        print('\n'+60*'-'+'\n')
```
